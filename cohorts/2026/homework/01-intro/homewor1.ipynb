```python
X = np.array(df[df["origin"] == 'Asia'][["vehicle_weight","model_year"]].reset_index(drop=True).loc[0:6])
```

```python
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
inv_XTX = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = inv_XTX.dot(X.T).dot(y) 
w.sum()
```
